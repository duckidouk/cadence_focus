```python
with sqlite3.connect(db_path) as conn:
    df = pd.read_sql_query(
        '''
        SELECT
            c.company_name,
            f.company_number,
            f.concept_local_name,
            f.value_numeric,
            f.unit,
            f.context_ref
        FROM financial_facts AS f
        LEFT JOIN company AS c
            ON f.company_number = c.company_number;
        ''',
        conn,
    )

df
```
