```python
import json
import sqlite3
from pathlib import Path


jlas = 'jlas-development-capital'
cadence = 'cadence-equity-partners'
scci = 'scci'
json_company_name = cadence
json_file_name = f"company_data_{json_company_name}-limited.json"

BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR / "data"
DB_PATH = BASE_DIR / "cf.db"

officer_list = []



with open(DATA_DIR / json_file_name) as f:
    data = json.load(f)
company_name = data["company_profile"]["company_name"]

for officer in data["officer_networks"]:

    officer_list.append(officer['officer_name'])

print(company_name)
print(officer_list)  
#print(data["officer_networks"])
#print(data.keys())

conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS company (
    company_name TEXT,
    company_number TEXT,
    officers TEXT
)

""")
```
